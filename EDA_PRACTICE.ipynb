```python
# Ensuring numeric columns are of correct types
numeric_cols = ['Age', 'Screen Time (hrs/day)', 'Sleep Duration (hrs)', 'Physical Activity (hrs/week)']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')
```

```python
df['Gender'] = df['Gender'].astype('category') # As Category takes less memory
```

```python
# Clean categorical columns (strip whitespace, unify capitalization)
cat_cols = ['Name', 'Gender', 'Stress Level', 'Anxious Before Exams', 'Academic Performance Change']
df[cat_cols] = df[cat_cols].apply(lambda x: x.str.strip().str.title())
```
