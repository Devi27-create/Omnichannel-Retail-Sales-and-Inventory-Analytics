```python
# Access a group of rows or columns in a pandas Dataframe
##the dataframe.loc is used in pandas to access a group of rows or columns in a dataframe. let's see an example.
import pandas as pd

#Dataset
data={
    'Student': ["Amit", "John", "Jacob", "David", "Steve"],
    'Rank': [1,2,3,4,5],
    'Marks': [95,70,80,60,90]
}

df=pd.DataFrame(data,index=['ROW A','ROW B','ROW C','ROW D', 'ROW E'])
print(f"Student Records: \n \n {df}")

#Accessing the value in the student column corressponsinf to row a lable.
a=df.loc['Row A', 'Student']
print(a)

```
