```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

##create a line plot
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Line Plot')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

## create a customized line plot 1

plt.plot(x,y, color='yellow', linestyle='-')
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

## create a customized line plot 2

plt.plot(x,y, color='pink', linestyle='--', marker='o', linewidth=3,markersize=6)
plt.grid(True)
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

## create a customized line plot 3

plt.plot(x,y, color='blue', linestyle='-.')
```

```python
## multiple plot

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1, color='green',marker='o')
plt.title('Plot 1')

plt.subplot(2,2,2)
plt.plot(y1,x, color='purple',marker='o')
plt.title('Plot 2')

plt.subplot(2,2,4)
plt.plot(x,y2, color='red',marker='o')
plt.title('Plot 4')

plt.subplot(2,2,3)
plt.plot(y2,x, color='black',marker='o')
plt.title('Plot 3')
```
