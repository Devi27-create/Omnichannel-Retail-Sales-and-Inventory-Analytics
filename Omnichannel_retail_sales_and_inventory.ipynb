```python
mask = (ord['Channel'].isin(['Online', 'App'])) & (ord['Store_Id'].isna())
ord.loc[mask, 'Store_Id'] = 'ONLINE'
ord.head(15)
```

```python
# Null Value Checked
ord.isnull().sum()
```

# After Correction Notes:
- Order_Date converted to date format. 
- Null value in Store_Id indeed null as channel for App and Online is not exist. So NaN in Store_Id filled by following rule: if NaN and Channel value is 'App' or 'Online', filled with 'ONLINE'.

```python
# Numpy Imported

import numpy as np
```

```python
mask2 = inv_move['Reason'].isin(['Restock', 'Return'])
inv_move['Change_Qty'] = np.where(mask2, inv_move['Change_Qty'].abs(),
                                  -inv_move['Change_Qty'].abs()
                                  )
inv_move.head(15)
```

## After Correction Notes:
- Date column converted to date format.
- change_qty already normalize using rule positive when reason column value is "Restock" or "Return", otherway make it negative.

# Data Cleaning

```python
## standardizing column names

prod.columns = prod.columns.str.strip().str.lower().str.replace(" ", "_")

ord.columns = ord.columns.str.strip().str.lower().str.replace(" ", "_")

cust.columns = cust.columns.str.strip().str.lower().str.replace(" ", "_")

ord_it.columns = ord_it.columns.str.strip().str.lower().str.replace(" ", "_")

inv.columns = inv.columns.str.strip().str.lower().str.replace(" ", "_")

inv_move.columns = inv_move.columns.str.strip().str.lower().str.replace(" ", "_")

stor =  stor.columns.str.strip().str.lower().str.replace(" ", "_")

ware = ware.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
# Store Table Text Cleaning

stor['city'] = stor['city'].str.strip().str.title()
stor['store_type'] = stor['store_type'].str.strip().str.title()
```

```python
# Duplicates Check

stor.duplicated()
```

```python
# Remove Duplicates

stor = stor.drop_duplicates()
cust = cust.drop_duplicates()
ord_it = ord_it.drop_duplicates()
```

## Relationships Checks

```python
# Store & orders relation check

ord['store_id'].isin(stor['store_id']).all()
```

```python
# Warehouse & inventroy relation check

inv['warehouse_id'].isin(ware['warehouse_id']).all()
```

```python
# Warehouse & Inventory Movement relation check

inv_move['warehouse_id'].isin(ware['warehouse_id']).all()
```

```python
# Removing Invalid Values

ord_it = ord_it[ord_it['quantity'] > 0]
```

```python
# Saving Clean CSVs

cust.to_csv("clean_customers.csv", index=False)
prod.to_csv("clean_products.csv", index=False)
ord.to_csv("clean_orders.csv", index=False)
ord_it.to_csv("clean_order_items.csv", index=False)
inv.to_csv("clean_inventory.csv", index=False)
inv_move.to_csv("clean_Inventory_movements.csv", index=False)
stor.to_csv("clean_stores.csv", index=False)
ware.to_csv("clean_warehouses.csv", index=False)
```
