```python
mask = (ord['Channel'].isin(['Online', 'App'])) & (ord['Store_Id'].isna())
ord.loc[mask, 'Store_Id'] = 'ONLINE'
ord.head(15)
```

```python
ord.isnull().sum()
```

Order_Date converted to date format. null value in Store_Id indeed null as channel for App and Online is not exist. So NaN in Store_Id filled by following rule: if NaN and Channel value is 'App' or 'Online', filled with 'ONLINE'.

### Notes:
- date dtype is wrong.
- Restock: stock comes IN, must be positive. Needs transformation.
- Sale: stock goes OUT, must be negative. Needs transformation.

```python
inv_move['Date'] = pd.to_datetime(inv_move['Date'], errors='coerce')
inv_move.info()
```

```python
inv_move.head(15)
```

```python
import numpy as np
```

```python
mask2 = inv_move['Reason'].isin(['Restock', 'Return'])
inv_move['Change_Qty'] = np.where(mask2, inv_move['Change_Qty'].abs(),
                                  -inv_move['Change_Qty'].abs()
                                  )
inv_move.head(15)
```

date column converted to date format. change_qty already normalize using rule positive when reason column value is "Restock" or "Return", otherway make it negative.
