# Data Cleaning

```python
## standardizing column names

prod.columns = prod.columns.str.strip().str.title()

ord.columns = ord.columns.str.strip().str.title()

cust.columns = cust.columns.str.strip().str.title()

ord_it.columns = ord_it.columns.str.strip().str.title()

inv.columns = inv.columns.str.strip().str.title()

inv_move.columns = inv_move.columns.str.strip().str.title()
```

```python
## Fix dtypes

prod['signup_date'] = pd.to_datetime(prod['signup_date'])

ord['order_date'] = pd.to_datetime(ord['order_date'])

inv_move['date'] = pd.to_datetime(inv_move['date'])
```
